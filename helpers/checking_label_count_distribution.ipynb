# Label Distribution Checker

This notebook checks the label count distribution of the `y` data CSV files (like `y_train.csv` and `y_test.csv`).

```python
import pandas as pd
import os
```

```python
def check_distribution(file_path):
    if not os.path.exists(file_path):
        print(f"Error: File '{file_path}' not found.")
        return

    try:
        df = pd.read_csv(file_path)
    except Exception as e:
        print(f"Error reading '{file_path}': {e}")
        return

    # Auto-detect label column
    if 'sleep_stage' in df.columns:
        target_col = 'sleep_stage'
    elif len(df.columns) == 1:
        target_col = df.columns[0]
    else:
        print(f"Error: Could not determine the label column in '{file_path}'.")
        return

    print(f"\n--- Label Distribution for: {file_path} ---")
    counts = df[target_col].value_counts().sort_index()
    proportions = df[target_col].value_counts(normalize=True).sort_index()

    STAGE_NAMES = {0: 'Wake', 1: 'N1', 2: 'N2', 3: 'N3', 5: 'REM'}

    print(f"{'Label (Code)':<15} | {'Count':<10} | {'Proportion'}")
    print("-" * 42)
    
    for val, count in counts.items():
        name = STAGE_NAMES.get(val, 'Unknown')
        label_str = f"{name} ({val})"
        prop = proportions[val]
        print(f"{label_str:<15} | {count:<10} | {prop:.2%}")
        
    print("-" * 42)
    print(f"{'Total':<15} | {counts.sum():<10} | 100.00%\n")
```

```python
# Check y_train.csv and y_test.csv in the root folder
files_to_check = ['../y_train.csv']

for file in files_to_check:
    check_distribution(file)
```
